## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_rows = 2500


df = pd.DataFrame({
    'page_url': [f"https://site.com/p_{i%50}" for i in range(n_rows)],
    'query': [f"term_{i%100}" for i in range(n_rows)],
    'month': ['2026-03'] * n_rows,
    'past_clicks': np.random.poisson(lam=20, size=n_rows),
    'past_impressions': np.random.poisson(lam=400, size=n_rows),
    'avg_position': np.random.uniform(1.0, 30.0, size=n_rows),
    'IS_TRUE': np.random.choice([True, False], size=n_rows, p=[0.90, 0.10])
})

df_valid = df[df['IS_TRUE'] == True].copy()

print("=== 3. QUERY VERIFICATION ===")
print(f"Total Rows Verified      : {len(df)}")
print(f"IS_TRUE Filtered Rows    : {len(df_valid)}")
print(f"Missing Values Count     : {df_valid.isnull().sum().sum()}")
print(f"Unique Page-Query Grains : {len(df_valid[['page_url', 'query']].drop_duplicates())}")
```
