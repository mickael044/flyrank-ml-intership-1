**1. Introduction / Problem Statement**
Content staleness leads to diminished user engagement. The goal of this research is to identify decaying content accurately using historical metrics without relying on manual audits, transitioning from a heuristic rule to a validated predictive model.

- **Features:** Staleness (days since update), CTR drop, and conversion rates.

## 3. Methodology

**3. Methodology & Validation Setup**

- **Model Selection:** Random Forest Classifier trained on historical content interaction features.

- **Leakage Prevention:** Future-window engagement metrics and post-action attributes were excluded prior to model training.
- **Evaluation Benchmark:** Compared directly against the Week 4 heuristic baseline (staleness > 30 days & CTR drop > 15%).

## 5. Limitations

*What this work cannot claim.*

**5. Limitations & Honest Framing**

- **Decision-Support Only:** The model output provides directional recommendations rather than absolute automated decisions.
- **Seasonal Sensitivity:** Temporary traffic drops due to seasonal trends might be misclassified as content decay.
- **Human In the Loop:** High-converting content requires mandatory manual review before executing any content refresh actions.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

**6. Actionable Playbook Recommendations**

1. **Prioritize High-Impact Refresh:** Execute `FULL_REFRESH` on content flagged with both high staleness (>45 days) and significant CTR drop (>20%).
2. **Title & Thumbnail Testing:** Apply light metadata updates for items showing sudden CTR drops without deep content decay.
3. **Monitoring & Drift Retraining:** Set up monthly Population Stability Index (PSI) checks; trigger automated model retraining if feature drift exceeds 0.25.
