## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

```python
**2. Data**
The analysis utilizes client interaction records spanning the last 90 days. Future-window metrics and post-action targets were explicitly excluded to prevent data leakage.

**3. Methodology**
- **Features:** Staleness (days since update), CTR drop, and conversion rates.
- **Baseline:** A heuristic rule triggering action if staleness > 30 days and CTR drop > 15%.
- **Model:** Random Forest Classifier.
- **Validation:** GroupKFold split by client ID to prevent cross-client data leakage.
```
