## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Binary Classification / Scoring

My chosen lane is Content Refresh Prioritization. From an ML perspective, this is a Binary Classification and Scoring task. For every page, we want to predict a binary outcome: Should this page be refreshed right now? (Yes/No - 1/0). At the same time, the model will output a probability score between 0 and 1, which allows us to rank the pages from "highest recovery potential" to "lowest," so editors can easily work down a prioritized list.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Since we don't have a perfect historical label saying exactly which page would have succeeded if updated, we use a proxy target based on historical data. Our target will be 1 (Yes) if a page has a high traffic decay trend (trend_direction == 'down') and a significant amount of lost clicks (clicks_lost > threshold), meaning it has a proven gap that can be recovered. It will be 0 (No) if the page is stable, growing, or losing very negligible traffic

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metrics: Precision @ K and ROC-AUC

    Business Metric: The percentage of recovered search traffic per editor hour spent.

    ML Metric: Precision @ K (e.g., Precision @ 100). Because editorial team time is limited, we want to make sure that if the model picks the top 100 pages to refresh, as many of them as possible actually turn out to be high-potential traffic winners. ROC-AUC will also help us evaluate how well the model separates decaying pages from healthy ones overall.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule like "Refresh any page that lost more than 500 clicks" sounds easy, but it fails in the real world. Traffic decay is complex—it depends on the page's age, its topic seasonality, historical volatility, and impressions. A fixed rule would cause a massive amount of False Positives (wasting hours on pages that are naturally dying or seasonal) and False Negatives (missing hidden gem pages that lost fewer clicks but are incredibly easy to fix). ML beats rules because it learns the subtle, multi-variable interactions between traffic patterns to rank pages accurately.
